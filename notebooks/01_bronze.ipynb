```python
from pyspark.sql import functions as f

CATALOG = "workspace"
SCHMEA = "hc_bronze"
RAW_PATH = f"/Volumes/{CATALOG}/{SCHMEA}/raw_files"

tables = ["patients","encounters","conditions","procedures","medications","providers","organizations","payers"]

for name in tables:
  df = (
      spark.read.csv(f"{RAW_PATH}/{name}.csv",header=True)
      .withColumn("_ingest_time",f.current_timestamp())
      .withColumn("_source_file",f.col("_metadata.file_path"))
  )
  df.write.mode("overwrite").saveAsTable(f"{CATALOG}.{SCHMEA}.{name}")
  print(name,"loaded:",df.count(),"rows")
```

```python
%sql
select count(*) from workspace.hc_bronze.encounters;
select _source_file,count(*) from workspace.hc_bronze.patients group by _source_file;
```
